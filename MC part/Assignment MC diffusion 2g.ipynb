```python
import numpy as np
import matplotlib.pyplot as plt
A = 0.8
N = 100
r = 1
x = 0
y = 0
X = [0]
Y = [0]
attempts = 0
for i in range(N):
    accepted = False
    while accepted == False:
        attempts += 1
        # Choose phi using g(phi)
        if np.random.rand() < (1 + A)/(2 + A):
            phi = np.pi*(np.random.rand() - 0.5)
            if phi < 0:
                phi = phi + 2*np.pi
            g = (1 + A)/(2*np.pi)
        else:
            phi = np.pi/2 + np.pi*np.random.rand()
            g = 1/(2*np.pi)
        P = (1 + A*np.cos(phi))/(2*np.pi)
        test = g*np.random.rand()
        if test <= P:
            accepted = True
    x = x + r*np.cos(phi)
    y = y + r*np.sin(phi)
    X.append(x)
    Y.append(y)
efficiency = N/attempts
print("Importance sampling efficiency =", efficiency)
plt.plot(X, Y)
plt.xlabel("x")
plt.ylabel("y")
plt.axis("equal")
plt.show()
```
